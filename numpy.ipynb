```python
import numpy as np
```

```python
arr1 = np.array([12,3,4,5])
print("arr1:", arr1)

print("arr1 shape:", arr1.shape)
```

```python
arr2 = np.array([[1,2,3],[4,5,6]])
print("arr2:", arr2)

print("arr2 shape:", arr2.shape)
```

```python
# zeros and ones
z = np.zeros((3,3))
o = np.ones((2,3))

print("z:", z)
print("o:", o)
```

```python
print(arr2.ndim)
print(arr1.ndim)
```

```python
arr4 = np.arange(1,11)
print(arr4)
```

```python
print(arr4.sum())
```

```python
# flatten
print(arr1.flatten())
print(arr2.flatten())
print(z.flatten())
print(arr4.flatten())
```

```python
# transpose T

arr5 = np.array([[1,2], [3,4], [5,6]])
print(arr5)
print(arr5.T)
```

```python
# dot() multiplication

arr6 = np.array([[1,2], [3,4]])
arr7 = np.array([[5,6], [7,8]])

print(arr6.dot(arr7))
```

```python
# random.rand
arr8 = np.random.rand(2,3)
print(arr8)
```
